```
def multi_Lp_dist(x, y):
    """
    Args:
      x: pytorch Variable, with shape [ns, k, d]
      y: pytorch Variable, with shape [ns, nc, d]
    Returns:
      dist: pytorch Variable, with shape [ns, nc, k]
    """
    p = 2
    ns, k, d = x.shape
    ns, nc, dy = y.shape
    assert d == dy  
    xx = x.view(ns, 1, k, d).expand(ns, nc, k, d)
    yy = y.view(ns, nc, 1, d).expand(ns, nc, k, d)
    dist = (torch.abs(xx-yy)**p).sum(dim=3).clamp(min=1e-12)**(1/p)
#    dist.addmm_(1, -2, x, y.t())
#    dist = dist.clamp(min=1e-12).sqrt()  # for numerical stability
    return dist
```

```
class OnlineClustering(nn.Module):

    """
    input principal:

        fv: feature vector en provenance de "contrastive model"
    """
    def __init__(self, 
                 averaging = {'method': 'global_mean',
                                            'param': None},
                 clusters=None,
                 thresh=10.):

        super(OnlineClustering, self).__init__()
        self.valmax = 1e3
        self.averaging = averaging
        if clusters is None:
            self.clusters = {} 
        else:
            self.clusters = clusters
        # clusters: {'mc': mc, shape (num_seq, nc_max, num_latentdims), 
        #           'sd':sdc, shape (num_seq, num_clusters, latent_dims),
        #           'nc':
        #           'size':sizec, shape (num_seq, 1), }
        self.distance = multi_Lp_dist
        self.nc_max = 0
        # threshold: seuil au dessus duquel: nouvel classe
        self.perceptron = nn.Sequential(
                                nn.Linear(1, 2, bias=False),
                                nn.Softmax(dim=1)
                                )
        # Init perceptron
        with torch.no_grad():
          self.perceptron[0].weight = \
                torch.nn.Parameter(torch.tensor([[-1.],[1.]]))
        self.raw_threshold = torch.nn.Parameter(torch.tensor([thresh]))

    def init_cluster(self, fv_init, nc_max, device='cuda:0'):

        #v0n de taille num_seq, num_latentdims, 
        # ns, size, *im_shape = images0.shape
        # v0 = self.cm(images0.reshape((ns*size,*im_shape)))
        # _, nld = v0.shape
        # v0 = v0.reshape(ns, size, nld)

        ns, size, nld = fv_init.shape
        self.nc_max = nc_max
        self.clusters['meanc'] = torch.zeros(ns, nc_max, nld).to(device)
        self.clusters['numc'] = torch.zeros((ns,)).to(device)
        self.clusters['sizec'] = torch.zeros((ns, nc_max)).to(device)
        self.clusters['mask_for_dist'] = torch.zeros((ns, nc_max, 1)).to(device)
        self.counter = torch.arange(self.nc_max).expand(ns, self.nc_max).to(device)

        

        #init with inputs:

        with torch.no_grad():             
            self.clusters['meanc'][:, 0, ...] = torch.mean(fv_init, dim=1)
            self.clusters['numc'] = torch.ones((ns,)).long().to(device)
            self.clusters['sizec'][:, 0] = size
            self.clusters['mask_for_dist'] = \
                 self.counter >= self.clusters['numc'].view(ns, 1)\
                                                      .expand(ns, nc_max)

            

    def forward(self, fv, device='cuda:0'): 
        # step 1: init assigned clusters
        with torch.no_grad():
            ns, nld = fv.shape    
            assigned_clusters = -1 * torch.ones(ns).long().to(device)
        # step 2: compute the distance, the min distances
            dist_to_centers = self.distance(fv.view(ns, 1, nld),
                                  self.clusters['meanc']).view(ns, self.nc_max)

            if (dist_to_centers >= self.valmax).sum() > 0:
                print(dist_to_centers)
                raise Exception('valmax pas assez haute')

            dist_to_centers = dist_to_centers \
                            + self.valmax * self.clusters['mask_for_dist'].float()
            mins, argmins = torch.min(dist_to_centers, dim=1)

        # step 3: construct a vector of proba to be in a new cluster : 
        # ns x 2 proba new cluster / proba no new cluster   # ns

        proba_newcluster = self.perceptron(mins.view((ns,1))\
                                           - self.raw_threshold
                                           )
        proba_clusters_0 = torch.zeros(ns, self.nc_max).to(device)
        proba_clusters_1 = torch.zeros(ns, self.nc_max).to(device)
        proba_clusters_0[range(ns), argmins] += proba_newcluster[:,0]
        proba_clusters_1[range(ns), self.clusters['numc'].clone()]\
                                   += proba_newcluster[:,1]

        proba_clusters = proba_clusters_0 + proba_clusters_1

        with torch.no_grad():
        # step 4: 
            # index of seq to change
            cond_newcluster = (proba_newcluster[:,1] > 0.5) 
            cond_oldcluster = (proba_newcluster[:,1] <= 0.5) 
            num_newcluster = cond_newcluster.sum()    
            num_oldcluster = cond_oldcluster.sum()
            
        # step 5: get the predicted labels : 
            assigned_clusters[cond_newcluster] = self.clusters['numc'][cond_newcluster]
            assigned_clusters[cond_oldcluster] = argmins[cond_oldcluster]

        # step 6: aggregate to the nearer cluster if min < threshold:
            # index of vectors to changes
            if num_oldcluster > 0:
                M = (torch.zeros(ns, self.nc_max) != 0).to(device)
                M[range(ns), argmins] = True   
                M[cond_newcluster,:] = False #normalement: torch.logical_not(cond_newcluster)
                if self.averaging['method'] == 'global_mean':
                  self.clusters['meanc'][M,:] *=\
                      self.clusters['sizec'][M].view(num_oldcluster,1)
                  
                  self.clusters['meanc'][M,:] += fv[cond_oldcluster,:] 
                  self.clusters['meanc'][M,:] /=\
                      (self.clusters['sizec'][M] + 1).view(num_oldcluster,1)
                elif self.buffersaveraging['method'] == 'exponential':
                  self.clusters['meanc'][M,:] *= (1 - averaging['param'])
                  self.clusters['meanc'][M,:] += averaging['param']\
                                                  * fv[cond_oldcluster,:] 

                elif self.averaging['method'] == 'mix':
                  coeff = 1./self.clusters['sizec'][M]\
                                 .view(num_oldcluster,1)
                  coeff = coeff.clip(min=averaging['param'])
                  self.clusters['meanc'][M,:] *= (1 - coeff)
                  self.clusters['meanc'][M,:] += coeff * fv[cond_oldcluster,:] 
                # update size of cluster:
                self.clusters['sizec'][M] += 1   
        # step 7: define new clusters if min > threshold:
            if num_newcluster > 0:
                M = self.counter == self.clusters['numc'].view(ns,1)\
                                                         .expand(ns, self.nc_max)
                M = M.to(device)
                M[cond_oldcluster,:] = False
                self.clusters['meanc'][M,:] = fv[cond_newcluster,:] 
                # print(cond_newcluster.long())
                self.clusters['sizec'][M] = 1
                self.clusters['numc'][cond_newcluster] += 1

        # update the mask for the distance:
            self.clusters['mask_for_dist'] = \
               self.counter >= self.clusters['numc'].view(ns,1)\
                                                    .expand(ns, self.nc_max)  

        # step 8: check that there is always self_nc_max > max(nc)
        # if not, update the shapes

        if torch.max(self.clusters['numc']) >= self.nc_max - 1:
            self.nc_max += 1
            self.clusters['meanc'] = torch.cat((self.clusters['meanc'],
                                         torch.zeros(ns, 1, nld).to(device)),
                                         dim=1) 
            self.clusters['sizec'] = torch.cat((self.clusters['sizec'],
                                         torch.zeros(ns, 1).to(device)),
                                         dim=1)
            self.clusters['mask_for_dist'] = \
                              torch.cat((self.clusters['mask_for_dist'],
                                         torch.ones(ns, 1).bool().to(device)),
                                        dim=1) 
            self.counter = torch.arange(self.nc_max).expand(ns, self.nc_max)\
                                                    .to(device)

        return  proba_newcluster, proba_clusters, assigned_clusters 


    # les "poids" sont chargés à la main, pourt l'instant
    def save_dict_clusters(self, path_dict_clusters):
        dict_clusters = {'clusters':self.clusters,
                           'nc_max':self.nc_max,
                           'counter':self.counter}
        torch.save(dict_clusters, path_dict_clusters)


    def load_dict_clusters(self, path_dict_clusters, device='cuda:0'):

        dict_clusters = torch.load(path_dict_clusters,
                                           map_location=device)

        self.nc_max = dict_clusters['nc_max']
        self.counter = dict_clusters['counter']
        self.clusters = dict_clusters['clusters']

```
